## DataFrame in Simple View
*Without a DataFrame, we will see the output like this:*

```python
import pandas as pd
names = ['United States', 'Australia', 'Japan', 'India', 'Russia', 'Morocco', 'Egypt']
dr =  [True, False, False, False, True, True, True]
cpc = [809, 731, 588, 18, 200, 70, 45]
my_dict = {'country':names,'drives_right':dr,'cars_per_cap':cpc}
print(my_dict)  
```

*With a DataFrame, we can see all the data neatly organized in a structured frame like this:*

```python
df = pd.DataFrame(my_dict)
print(df)
```

*Instead of the default numeric index (0 to 6), I changed the index to country short codes to make the data more readable and meaningful:*

```python
df.index = ['US', 'AUS', 'JPN', 'IND', 'RUS', 'MOR', 'EGY']
print(df)
```
