# World Population Data Overview
I downloaded this CSV file from **Kaggle – World Population 2026**.

```python
import matplotlib.pyplot as plt
from pandas import read_csv
data = read_csv('C:\\Users\\user\\OneDrive\\VS CODE\\DATA_VIZULATION\\archive\\world_population_by_country_2026.csv')
print(data.head(5))
```

## Using a bar chart for clarity
I created a small bar chart to visualize the **top 10 most populated countries** in 2026.

```python
top10 = data.head(10)
plt.figure(figsize=(10,6))
plt.bar(top10['Country'], top10['Population_2026'], color='skyblue')
```

## Short labels for clarity

```python
plt.bar(top10['Country'], top10['Population_2026'], color='skyblue')
#This bar chart visualizes the top 10 most populated countries in 2026.
#Using abbreviations (e.g., 'IND' instead of 'India') on the x-axis
#This makes the chart cleaner and easier to read, especially when country names are long
plt.xticks(['India', 'China', 'United States', 'Indonesia', 'Pakistan', 'Nigeria','Brazil', 'Bangladesh', 'Russia', 'Ethiopia'], ['IND', 'CHN', 'USA', 'IDN', 'PAK', 'BRA', 'NGA', 'BGD', 'RUS', 'ETH'])
#X-axis: Country names (shortened labels for clarity)
plt.xlabel('Country')
#Y-axis: Population in 2026
plt.ylabel('Population in 2026')
#Title: Top 10 Most Populated Countries in 2026
plt.title('Top 10 Most Populated Countries in 2026')
plt.show()
```
